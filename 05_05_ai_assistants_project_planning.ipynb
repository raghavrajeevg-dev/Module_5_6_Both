## 1. What AI assistants can and cannot do (10 min)

AI assistants can help explain code, suggest data checks, draft report wording, brainstorm features, and improve presentation structure. They can also make mistakes, invent sources, expose confidential data if used carelessly, or give overconfident advice.

Rules for responsible use: do not paste confidential plant data into unapproved tools; verify claims and calculations; state when AI assisted work; retain human ownership of engineering judgement.

```python
import pandas as pd

case_data = pd.read_csv('https://raw.githubusercontent.com/raghavrajeevg-dev/Module_5_6_Both/main/data/quality_batches_120.csv')
case_data['timestamp'] = pd.to_datetime(case_data['timestamp'])
case_data.head()
```

## 2. Frame a project question (12 min)
A good student project is narrow, measurable, and tied to a user.

Example: Use historic batch data to identify records that may require quality review.

Define user, decision supported, CSV source, target/outcome, input features, success measure, limits, and required human review. Avoid claiming that a short classroom dataset can validate a production AI system.

```python
project_snapshot = case_data.groupby('lab_result').agg(
    records=('batch_id', 'count'),
    mean_conversion=('conversion_percent', 'mean'),
    mean_purity=('product_purity_percent', 'mean')
)
display(project_snapshot.round(2))
print('Potential input features:', ['inlet_temperature_c', 'conversion_percent', 'reactor'])
print('Potential outcome label: lab_result')
```

Use only non-confidential or approved data. Run suggested code yourself and inspect results. Treat suggested code as a draft requiring review.
